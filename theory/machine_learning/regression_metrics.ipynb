# Regression Metrics 

## MAE - Mean Absolute Error 

Indicate that how far the predict value from the actual value in **overall**

Suppose that we have this table 

| Actual (y_true) | Predicted (y_pred) | Error (y_pred − y_true) |
| --------------- | ------------------ | ----------------------- |
| 3               | 2.5                | −0.5                    |
| −0.5            | 0.0                | +0.5                    |
| 2               | 2                  | 0                       |
| 7               | 8                  | +1                      |

The **error** indicate how far of the predict value of individual within the actual value, so we have 

['-0.5', '0.5', '0', '1']

To calculate in overall, we have a formula that 

$$
MAE = \frac{1}{n} \sum |y_{true} - y_{pred}|
$$

In above example, the MAE would be **0.5** which indicate that the prediction is off the actual data **0.5 units**

## MSE - Mean Squared Error (punishes bug mistakes)

So basically we calculate the squared of each error and find the average between all of them 

The formula should be 

$$
MSE = \frac{1}{n} \sum (y_{true} - y_{pred})^2
$$

The reason that this metric use to punish the mistake, it actually is not a real punishment, it is more about we need to be careful with the big error which it can really make a huge impact 

## RMSE - Root Mean Squared Error (MSE but human-friendly)

RMSE is used to calculate how far from the actual data, which is the same with MAE but more precise as it take a deeply consideration about the big error. The formula should be 

$$
RMSE = \sqrt{MSE}
$$

This doesn't mean RMSE is better than MAE, we can think that we use RMSE when the big error become a problem and use MAE when the error is nearly equal 

For example, we have 2 list of errors parameters 
1. [2,3,4,2,1] -> We can use MAE because we don't have a dramatical error 
2. [2,3,12,20,40] -> We should consider about RMSE as in this case, we need to take care about the big error 

## R^2 - "Is my model better than guessing"

R^2 compares the model to a baseline model (A model that predicts the mean of y for everything) 

If the model is not better than that -> It's useless 

An intuition: 
- R^2 = 1 -> perfect predictions 
- R^2 = 0 -> same as predicting the mean 
- R^2 < 0 -> worse than predicting the mean
