## NumPy - Mathematical & Statistical Operations


### Elementwise operations

A mathematical operations applied independently to each element of one or more arrays. NumPy performs these operations in **vectorized** form 

For example,


````python
import numpy as np 

a = np.array([1,2,3])
b = np.array([4,5,6])

result = a + b 
print(result)
print(type(result))
````

Where we defined 2 arrays and add them with corresponding row and column number. There are still more elementwise operations, which are

````python
numpy_add = np.add(a,b)

print(numpy_add)
print(type(numpy_add))
````

````python
multi_result = np.multiply(a, b)
print(multi_result)
````

````python
power_result = np.power(a, b)
print(power_result)
````

````python
sqrt_result = np.sqrt(a)
print(sqrt_result)
````

````python
log_result = np.log(a)
print(log_result)
````

````python
exp_result = np.exp(a)
print(exp_result)
````

### Reduction Operation (Aggregation)

Reduction operation collapse an array into a single value or fewer values, usually by summing, averaging, or finding min/max. NumPy implements these as highly optimized routines

````python
arr = np.array([1,2,3,4,5])
print(arr.sum())
print(arr.mean())
print(arr.std()) # Calculate the standard deviation
print(arr.var())
print(arr.min())
print(arr.max())
````

### Axis-based operations 

Axis-based operations perform reductions **along a specific dimension** of an array.

````python
A = np.array([[1,2,3], [4,5,6]])

print(A.sum(axis=0)) # sum by row
print(A.sum(axis=1)) # sum by column 
````

#### Different kind of implementation

We can use function API or method API 

````python
np.sum(A, axis=0)

#or 

A.sum(axis=0)
````

### Sorting & Ranking

Sorting arranges array elements in ascending order. Whereas, ranking returns the indices that would sort the array

````python
arr = np.array([30,10,20])

print(np.sort(arr))
print(np.argsort(arr)) # return the original index of each value before sorting
````

#### Different sorting methods 

1. `np.sort` (return sorted copy)

````python
new_array = np.sort(arr)
print(new_array)
````

2. `arr.sort()` (sorts in place)

````python
arr.sort()
print(arr)
````

### Statistical functions 

Statistical functions summarize distribution characteristics of data: mean, median, quantile, percentile, correlation,...

````python
arr = np.array([1,2,3,4,5])
print(np.median(arr))
print(np.percentile(arr, 75))
````

So the `percentile` here is a bit tricky to calculate it, it is following within this formula 

```
(P/100) * (N - 1)
```

where 
- P is the number of percentile that define in the code, in this case, it is 75 
- N is the number of element in an array 

So in above example, we have 

(75/100) * (5-1) = 4 

#### Different statistical methods 


````python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])

print(np.quantile(arr, 0.25))
print(np.corrcoef(arr1, arr2))
print(np.cov(arr1, arr2))
````

So the formula to calculate the quantile is kind of the same with percentage 

```
position = q * (N - 1)
```

Where **q** is the value specific in the code 

And the covariance and correlation coefficient is the mathematical formula 

### Linear Algebra

NumPy's `linalg` module implements matrix operations used in ML, optimization, and scientific computation.

````python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A @ B)  # matrix multiplication
print(A.T) # Transpose
print(np.linalg.inv(A)) # Inverse
print(np.linalg.det(A)) # Determinant 
print(np.linalg.eig(A))
````
