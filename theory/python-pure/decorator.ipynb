## Decorator 

A decorator is a function that 
- Taking another function as input 
- adds extra behavior before or after the function runs 
- returns a modified version of that function 

The decorators are built on 2 things 
- First-class functions -> Functions can be passed as argument 
- Closures -> inner functions can remember variables even after the outer function returns 

For example, 

```python
def my_decorator(func):
    def wrapper():
        print("Before calling function")
        func()
        print("After calling function")
    return wrapper

@my_decorator
def greet():
    print("Hello!")

greet()
```

So basically, the function above will work like this 

```python
def greet():
    print("Hello!")

greet = my_decorator(greet)

print(greet)
```

## Some real world example

### Logging

```python
def log(func):
    def wrapper(*args, **kwargs):
        print(f"[LOG] Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log
def addNumber(a,b): 
    return a + b 


addNumber(2,3)
```

### Execution timer

```python
import time
from functools import *

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Execution: {end - start:.4f}s")
        return result
    return wrapper

@timer
def mulNumber(*args):
    return reduce(lambda a,b: a*b, args)

mulNumber(2,3,4)
```

### Caching / Memorization

```python
def cached(func):
    cache = {}
    def wrapper(x):
        if x not in cache:
            cache[x] = func(x)
        return cache[x]
    return wrapper
```
