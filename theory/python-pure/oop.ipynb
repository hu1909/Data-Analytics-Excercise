## OOP (Object-Oriented Programming)

A programming paradigm based on object - bundles of data (attributes) and behaviors (methods)

There are different concepts are used to enhance the meaning of OOP     

### Classes & Objects 

A **class** is a blueprint 

An **object** is an instance of the blueprint

For example,

```python
class Animal: 
    def __init__(self, name, legs):
        self.name = name 
        self.legs = legs 

    
    def make_sound(self):
        print(f"{self.name} has just made sound and he has {self.legs}")

dog = Animal("Butter", 4)
dog.make_sound()
    
```

Within this function: 
- *dog* here is an instance of class **Animal** 
- The keyword `self` indicates an object on which a method is called

Inside the function, we can see like `self.name` or `self.legs` -> this means that an attribute belonging to the current object

### Encapsulation (Controlling Data Access)

Encapsulation = bundling data + methods, AND controlling access to data. Python use naming convention to enfore the privacy of a variable


#### Public

```python
self.name
```

#### Protected 

```python
self._salary 
```

#### Private

```python
self.__password
```

Below is an example about the 

```python
class Account: 
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance # protected 
        self.__pin = "1234" # private

hung_account = Account("Hung", 10000)
print(hung_account.owner)
print(hung_account._balance)
# print(hung_account.__pin) # This won't work 
print(hung_account._Account__pin) # Not recommended 
```

### Abstraction (Hide implementation details)

Abstraction means that we will hide how something works internally

For example, 

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass 
```

### Inheritance (Reuse and Extend Classes)

A class can inherit from another class, and there are multiple types of inheritance in python 
- Single inheritance 
- Multiple inheritance 
- Multilevel inheritance 
- Hierarchical inheritance

For example,

```python
class Animal:
    def speak(self):
        print("Some sound")
    
class Dog(Animal):
    def speak(self):
        print("Woof!")
    
d = Dog()
d.speak()
```

The `super()` function used to call parent class methods

```python
class Animal: 
    def __init__(self, legs):
        self.legs = legs

class Dog(Animal):
    def __init__(self, legs, name):
        super().__init__(legs)
        self.name = name 
```

### Polymorphism (Same method name - Different behaviors)

**Example 1: Method Overriding**

```python
class Animal:
    def speak(self):
        print("Animal sound")

class Cat(Animal):
    def speak(self):
        print("Meow")
```

**Example 2: Duck Typing**

```python
class Cat:
    def speak(self):
        print("Meow")

class Human:
    def speak(self):
        print("Hello")

def make_sound(obj):
    obj.speak()

make_sound(Cat())   # Meow
make_sound(Human()) # Hello
```

### Special "Magic Methods"

These methods let our class behave like Python built-ins 

- `__init__` - constructor 
- `__str__` - user-friendly string
- `__repr__` - developer debug output
- `__getattr__` - called only when attribute is not found
- `__getattribute__` - cllaed for all atrribute access 
- `__len__` - define the behavior of **len(obj)**
- `__contains__` - called when using **in** operator 
- `__iter__` - return an iterator 
- `__getitem__` - makes object subscriptive **obj[key]**
- `__new__` - the method responsible for creating the object in memory, before init run so it is suitable for singleton pattern (singleton pattern is that we only let 1 instance run at a time)

```python
class Vector:

    def __new__(cls):
        if cls.instance is None: 
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, *values):
        self.values = list(values)

    def __repr__(self):
        return f"Vector({', '.join(map(str, self.values))})"

    def __str__(self):
        return f"<{', '.join(map(str, self.values))}>"

    def __len__(self):
        return len(self.values)

    def __getitem__(self, index):
        return self.values[index]

    def __add__(self, other):
        return Vector(*[a+b for a,b in zip(self.values, other.values)])

    def __eq__(self, other):
        return self.values == other.values

    def __iter__(self):
        return iter(self.values)

v1 = Vector(1,2,3)
v2 = Vector(4,5,6)

print(v1 + v2)       # <5, 7, 9>
print(len(v1))       # 3
print(v1[1])         # 2
print([x for x in v1]) # [1,2,3]
```

### Class Mathods & Static Methods & Abstract Methods & Instance method

#### Instance method 

Has access to **self** which mostly use to declare the behavior of an instance

```python
def method(self):
    pass
```

#### Class method 

Uses `@classmethod` and receives `cls`, this method usually belong to class-level which manipulates class-level attributes

```python
class A:
    counter = 0
    @classmethod
    def increase(cls):
        cls.counter += 1
```

#### Abstract method 

Uses `@abstractmethod`, this's mostly defined abstract class and only provide a basic structure not a fully implementation 

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass 
```

#### Static method 

Uses `@staticmethod`, this method is independent with a class but usually used as a helper for the main function 

```python
class Math:
    @staticmethod
    def add(a, b):
        return a + b
```

### Property Decorator (Python-specific feature)

Used to create getters/setters in a clean way

```python
class Person:
    def __init__(self, age):
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Invalid age")
        self._age = value


p = Person(20)
p.age = 25
print(f"The age of this person is {p.age}")
```
