## Basic components in Python

### Order and Mutable 
- **Order** means the value in an object is indexing from 0 -> n (length of object)
- **Mutable** means the value in an object is accessible or not, which we can modify, remove or add new value

### List 

List is an **ordered** and **mutable** collection of items. For example, 


```python
empty_list = []
string_list = ["Banana", "Apple", "Strawberry"]
numeric_list = [1,2,3,4,5]
mix_list = ["Banana", 1, 2, "Apple"]

print(empty_list)
print(string_list)
print(numeric_list)
print(mix_list)
```

#### Append value into a list 

```python
empty_list.append(0)

print(empty_list)
```

#### Modify a value in a list 

```python
string_list[1] = "Blueberry"

print(string_list)
```

#### Return the number of occurence of specific value

```python
num_occurence = string_list.count("Banana")

print(f"Number of occurrence of Banana in a string: {num_occurence}")
```

#### Sort a list 

```python
sorted_list = sorted(string_list)

print(sorted_list)
```

#### Slicing a list 

```python
slicing_list = string_list[1:2]

print(slicing_list)
```

#### Reversing a list 

```python
reversed_list = sorted(string_list, reverse=True)

print(reversed_list)
```

### Tuple 

Tuple is an **ordered** and **immutable** collection of items. For example, 

```python
empty_tuple = ()

string_tuple = ("Banana", "Apple", "Strawberry")

number_tuple = (1,2,3,4)

print(empty_tuple)
print(string_tuple)
print(number_tuple)
```

#### Accesing a tuple 

```python
first_element = string_tuple[0]
print(first_element)

slice_tuple = string_tuple[:1]
print(slice_tuple)
```

#### Unpack a tuple 

```python
x, y, z = string_tuple 

print(x)
print(y)
print(z)
```

#### Count number of appearance of values

```python
occurrence_number = string_tuple.count('Apple')

print("Number of appearance of Apple: ", occurrence_number)
```

#### Return index of specific value

```python
index_of_apple = string_tuple.index('Apple')

print(f"The index of Apple value: {index_of_apple}")
```

### Set 

A set is **unordered** and **mutable** collection of **unique** items -> This means that we don't have duplicates 

But when **a set is created**, the element inside is unchangable where we can only add or remove elements 

For example,

```python
empty_set = {}

fruits = {"apple", "banana", "cherry"}

numbers_set = set(numeric_list)
print(numbers_set)
```

#### Add a single item

```python
fruits.add("orange")
print(fruits)
```

#### Add multiple items from an iterable

```python
fruits.update(["cherry", "strawberry"])

print(fruits)
```

#### Remove an item, Raises a KeyError if the item is not found

```python
try:
    fruits.remove("potato")
except KeyError:
    print("The value does not exist")

fruits.remove("cherry")
print(fruits)
```

#### Remove an item, but does nothing if the item is not found

```python
fruits.discard("banana")
fruits.discard("cherry")

print(fruits)
```

### Dictionary 

A dictionary is **mutable** and **ordered** collection of items that stores data in key-value pairs 

The key properties: 
- **Unique keys**: Keys must be unique within a dictionary. If we assign a value to an existing key, it will overwrite the old value 
- **Mutable**: We can *add*, *change*, or *remove* key-value pairs after the dictionary is created 
- **Ordered**: The dictionaries remember the order in which items were inserted

For example,

```python
user_infor = {
    "Name": "Hung", 
    "age": 24,
    "education": "Master Degree"
}

empty_dict = {}

print(user_infor)
print(empty_dict)
```

#### Accessing values

```python
print(user_infor["Name"])
print(user_infor["age"])

# or 

print(f"Another way to get age number: {user_infor.get("age")}")
```

#### Add or Modify Items

```python
# Add an item 
user_infor["job"] = "DevOps Engineer"

# Modify an item 
user_infor["age"] = 30 
print(f"My age after 6 years is: {user_infor["age"]}")
print(user_infor)
```

#### Remove items 

There are severals way: 
- **pop(key)** removes the item with specified key and return its value 
- **del** removes the items with the specified key 
- **clear()** removes all items from the dictionary

```python
remove_job = user_infor.pop("job")

print(remove_job)

del user_infor["age"]

print(user_infor)
```

#### Loop through elements

```python
for key in user_infor: 
    print(f"Here is the key: {key}")

for value in user_infor.values():
    print(f"Here is the value in a dictionary: {value}")


for key, value in user_infor.items():
    print(f"Here is {key} and corresponding value is {value}")
```
