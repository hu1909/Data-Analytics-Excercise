# Pandas: Indexing, Selection, Filtering, Cleaning, and Feature Preparation

In this notebook, we cover essential Pandas skills used in real-world data manipulation: 
- Selecting rows and columns using `.loc` and `.iloc`
- Boolean filtering and conditional selection 
- Handling missing values 
- Removing duplicates 
- Coverting data types 
- Creating new columns (feature engineering)
- Applying functions using `.apply`

These operations from the backbone of data cleaning and preparation before analysis or ML

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_rows", 10)
pd.set_option("display.max_columns", 20)

df = pd.read_csv("data/users.csv")
df.head()
```

## Indexing & Selection 


### Column Selection 

There are 3 common ways to select column: 
- Square brackets for a single column -> returns a **Series**
- Double brackets for multiple columns -> returns a **DataFrame** 
- Dot notation for simple column names  

```python
df["name"]
```

```python
df[["name", "age"]]
```

```python
df.city
```

### Row selection with .loc and .iloc 

1. **.loc** (label-based)
- Access rows/columns by **labels**
- Syntax: `df.loc[row_labels, column_labels]`

```python
df.loc[0]
```

```python
df.loc[0, ["name", "age"]] # Return specific values from the first row
```

```python
df.loc[0:5, ["name", "age", "city"]]
```

2. **.iloc** (position-based)
- Access rows/columns by **integer position** 
- Syntax: `df.iloc[row_positions, column_positions]`

```python
df.iloc[0]
```

```python
df.iloc[0, 0]
```

```python
df.iloc[10:20, 1:4] # row from 10 to 20 and column from 1 to 4 
```

## Boolean Filtering

Filtering rows with Boolean conditions 

Common operators: 
- `==`, `!=`
- `>`, `<`, `>=`, `<=`
- Combine with: 
    - `&` (AND)
    - `|` (OR)
    - `~` (NOT)

```python
df[df["name"] == "Alice" ]
```

```python
df[df["age"] > 25]
```

## Assignment & Modifying Data 

Always use `.loc` to safely set values 
Avoid chained indexing like `df[df["age"] >30]["salary"] = ....` because it may not modify the original DataFrame

```python
df.loc[df["city"] == "New York", "city"] = "Viet Nam"
df.head()
```

## Handling Missing Data   

Key functions: 
- `df.isna()` -> True/False mask 
- `df.isna().sum()` -> count missing per column 
- `df.dropna()` -> remove rows with missing values 
- `df.fillna(x)` -> fill missing values

```python
df.isna().sum()
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df_drop = df.dropna()
df_drop
```

## Handing Duplicates 

- `df.duplicated()` -> Boolean mask of duplicates 
- `df.drop_duplicates()` -> Remove duplicate rows 

```python
df.duplicated()
```

```python
df_no_dup = df.drop_duplicates()
df_no_dup.shape
```

## Type Conversion 

Useful methods: 
- `df["col"].astype("int64")`
- `pd.to_datetime(df["date_col"])`

```python
df["age"] = df["age"].astype("int64")
df["city"] = df["city"].astype("category")
df.dtypes
```

## Feature Engineering 

Creating new columns using vectorized operations: 
- Arithmetic 
- Conditional columns 
- `.apply()` for custom logic 

```python
def check_age(age):
    if age >= 25 and age <= 30:
        return "Young"
    elif age > 30:
        return "Old"
    else: 
        return "Teenager"

df["generation"] = df["age"].apply(check_age)
df[["age", "generation"]].head()

```

```python
df["age_squared"] = df["age"].apply(lambda x: x * x)
df[["age", "generation", "age_squared"]].head()
```
