# Panda Fundmentals: Series, DataFrames, I/O and Inspection 

## What is Panda ? 

Pandas is a Python library for **data manipulation and analysis**, build on top of NumPy. 

Key points: 
- Provides 2 main data structures: `Series` (1D) and `DataFrame` (2D)
- Designed for working with **tabular data** (like Excel sheets, CSV files, SQL tables)

In real projects, the typical flow is: 

Raw data -> Pandas for cleaning & transformation -> NumPy/sklearn for modeling

## Series 

A **Series** is a one-dimensional labeled array. 

Think of it as: 
- A single column of a spreadsheet, with an **index** (row labels) and **values**
- Similar to a NumPy array, but with labels and more functionality.

For example,

```python
import pandas as pd 

s1 = pd.Series([10,20,30,40])
s1
```

### Series with custom index 


```python
s2 = pd.Series(
    data=[10,20,30],
    index=["a", "b", "c"],
    name="example_series"
)

print("Series values: ", s2.values)
print("Series index: ", s2.index)
print("Series dtype: ", s2.dtype)
```

### Series from dictionary

```python
data_dict = {"apple": 3, "banana": 5, "orange": 2}
s3 = pd.Series(data_dict, name="fruit_count")
s3
```

### Notes on Series 

- `index`: labels for the elements (can be strings, numbers, dates, etc..)
- `values`: the underlying NumPy array
- `dtype`: data type of the elements (int, float, object,...)

We can access elements by label or position: 
- `s2["a"]` -> by label 
- `s2[0]` -> by position

```python
print("s2['a']  ->", s2["a"])
print("s2[0]    ->", s2[0])

print("\nSlice by position: ")
print(s1[1:3])
```

## DataFrames 

A **DataFrame** is a two-dimensional, tabular data structure with: 
- **Row** and **columns** (like an Excel sheet)
- Each column is a `Series`
- Row labels (index) and column labels (column names)

This is the main structure we will work with in Pandas 

### Create DataFrame from dicts 

```python
# dictionary 
data = {
    "name": ["John", "Anna", "Peter", "Linda"],
    "age": [23, 25, 31, 29],
    "city": ["HCM", "Hanoi", "Danang", "HCM"]
}

# List of dictionaries 
records = [
    {"name": "John", "age": 23, "city": "HCM"},
    {"name": "Anna", "age": 25, "city": "Hanoi"},
    {"name": "Peter", "age": 31, "city": "Danang"},
]

df_people = pd.DataFrame(data)
df_people
```

```python
df_people_v2 = pd.DataFrame(records)
df_people_v2
```

### Basic attributes of DataFrame

```python
print("Shape (rows, columns): ", df_people.shape)
print("Columns: ", df_people.columns.tolist())
print("Index: ", df_people.index)
print("\nData types: ")
print(df_people.dtypes)
```

### Reading Data from CSV

```python
customers = pd.read_csv("data/customers-100.csv")

customers.head()
```

#### Basic inspection 

```python
print("Shape: ", customers.shape)
print("\nColumns: ", customers.columns.tolist())

print("\nInfor:")
print(customers.info())

print("\nDescribe (numeric columns):")
print(customers.describe())
```

### Random sampling and tail 

- `.sample(n)` -> sample `n` random rows.
- `.tail()` -> last 5 rows (by default)

```python
print("Random 3 rows:")
display(customers.sample(3))

print("\nLast 5 rows:")
display(customers.tail())
```
